```python
import os
import base64
from openai import AzureOpenAI
from dotenv import load_dotenv

load_dotenv()

endpoint = os.getenv("ENDPOINT_URL")
deployment = os.getenv("DEPLOYMENT_NAME")
subscription_key = os.getenv("AZURE_OPENAI_API_KEY")
```

```python
# Create client

client = AzureOpenAI(
    api_key=subscription_key,
    azure_endpoint=endpoint,
    api_version="2025-01-01-preview"
)
```

```python
# Call the chat completion
response = client.chat.completions.create(
    model=deployment,
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "What is the capital of France?"}
    ],
    max_tokens=50,
    temperature=0.7
)

# Print the response
print(response.choices[0].message.content)
```

```python
def get_response(prompt):
    response = client.chat.completions.create(
        model=deployment,
        messages=[
            {"role": "system", "content": "You are a helpful assistant."},
            {"role": "user", "content": prompt}
        ],
        max_tokens=500,
        temperature=0.3
    )
    return response.choices[0].message.content
```

```python
prompt = "Give me a shopping list for pasta bolognese. Include ingredients and quantities. Use only bulletpoints."


response = get_response(prompt)
print(response)
```

### Zero-Shot Prompting

```python
prompt = "Which players do you think will be selected to NBA All-Star Game in 2025?"

response = get_response(prompt)
print(response)
```

### Few-Shot Prompting

```python
prompt = """Which players do you think will be selected to NBA All-Star Game in 2025?
Include a list of players with their teams and positions. Use bullet points for each player.
Give only position, name and team, do not include any other information.
For example:
West:
PG: Stephen Curry (Golden State Warriors)
SG: Devin Booker (Phoenix Suns)
...

East:
PG: Trae Young (Atlanta Hawks)
...
"""

response = get_response(prompt)
print(response)
```

```python
prompt = """Which players do you think will be selected to NBA All-Star Game in 2025?
Include a list of players with their teams and positions
Give only position, name and team, do not include any other information.
Return results in a table format.
"""

response = get_response(prompt)
print(response)
```
