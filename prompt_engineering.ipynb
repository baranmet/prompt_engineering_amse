```python
import os
from openai import AzureOpenAI
from dotenv import load_dotenv

load_dotenv()

endpoint = os.getenv("ENDPOINT_URL")
deployment = os.getenv("DEPLOYMENT_NAME")
subscription_key = os.getenv("AZURE_OPENAI_API_KEY")
```

```python
# Create client

client = AzureOpenAI(
    api_key=subscription_key,
    azure_endpoint=endpoint,
    api_version="2025-01-01-preview"
)
```

```python
# Call the chat completion
response = client.chat.completions.create(
    model=deployment,
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "What is the capital of France?"}
    ],
    max_tokens=50,  # Length of the response
    temperature=0.7  # Controls randomness in the response
)

# Print the response
print(response.choices[0].message.content)
```

```python
response = client.chat.completions.create(
    model=deployment,
    messages=[
        # 🧠 System role sets the LLM's behavior
        {"role": "system", 
         "content": "You are a genius football manager for a professional team."},

        # 🙋 User asks a question
        {"role": "user", 
         "content": """Whom should I focus on for our Champions League match 
         against Real Madrid?"""},

        # 🤖 Assistant gives a first response (optional but good for multi-turn)
        {"role": "assistant", 
         "content": """Vinicius Jr. is a key player for Real Madrid. 
         You should focus on containing his pace and dribbling."""},

        # 🙋 User follows up
        {"role": "user", 
         "content": "Anyone else I should consider?"}
    ]
)

print(response.choices[0].message.content)
```

```python
# Call the chat completion
response = client.chat.completions.create(
    model=deployment,
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "What is the capital of France?"}
    ],
    max_tokens=30,
    temperature=2
)

# Print the response
print(response.choices[0].message.content)
```

```python
def get_response(prompt):
    response = client.chat.completions.create(
        model=deployment,
        messages=[
            {"role": "system", "content": "You are a helpful assistant."},
            {"role": "user", "content": prompt}
        ],
        max_tokens=500,
        temperature=0.3
    )
    return response.choices[0].message.content
```

```python
prompt = "Give me a shopping list for pasta bolognese. Include ingredients and quantities. Use only bulletpoints."

response = get_response(prompt)
print(response)
```

### Zero-Shot Prompting

```python
prompt = "Which players do you think will be selected to NBA All-Star Game in 2025?"

response = get_response(prompt)
print(response)
```

### Few-Shot Prompting

```python
prompt = """Which players do you think will be selected to NBA All-Star Game in 2025?
Include a list of players with their teams and positions. Use bullet points for each player.
Give only position, name and team, do not include any other information.
For example:
West:
PG: Stephen Curry (Golden State Warriors)
SG: Devin Booker (Phoenix Suns)
...

East:
PG: Trae Young (Atlanta Hawks)
...
"""

response = get_response(prompt)
print(response)
```

```python
prompt = """Which players do you think will be selected to NBA All-Star Game in 2025?
Include a list of players with their teams and positions
Give only position, name and team, do not include any other information.
Return results in a table format.
"""

response = get_response(prompt)
print(response)
```

```python
prompt = """
You are a talented Dutch musician and songwriter, deeply inspired by the heartfelt and relatable style of the legendary André Hazes.
Please help me write lyrics for a song that will be performed at the event.
Axians Microsoft Summer Event is an internal, semiformal activity where Axians colleagues meet each other and share expertise.
I need two verses of lyrics. Each verse should be 4–6 lines long.
The tone should be warm, friendly, and nostalgic—true to André Hazes’ signature style. Feel free to include references to summer, teamwork, and the joy of being together.
"""

response = get_response(prompt)
print(response)
```

```python
prompt = """
I want you to  help me write a Python code. Please follow the following steps:
1. Write a function that takes an argument and multiples it by 2.
2. Use this function in a series of numbers that range from 1 to 5. 
3. Return the results in a list.
4. Add all the results together and return the sum.
"""
response = get_response(prompt)
print(response)
```
